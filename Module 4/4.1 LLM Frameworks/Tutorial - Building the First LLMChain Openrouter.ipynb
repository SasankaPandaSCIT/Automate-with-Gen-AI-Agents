## Tutorial: Building the First LLMChain and Understanding Its Structure 

In this short hands-on, we turn a single prompt into a reusable chain you can run locally. We’ll keep it minimal and production-minded: clear inputs, deterministic defaults, and environment-safe keys.

Learning outcomes:
- Understand what a `PromptTemplate` is and why it matters
- Wrap a prompt + model into an `LLMChain`
- Execute the chain with variables
- Add small improvements for reliability (temperature, seed, stop tokens)

Prereqs: an OpenAI-compatible key in your environment (e.g., `OPENAI_API_KEY`). If you don’t have one, you can still read the code and plug a different provider later.


### Step 1: Set up environment and imports
We isolate configuration from code. Put secrets in `.env` or environment variables, not notebooks.


```python
from getpass import getpass



# OpenRouter configuration
from getpass import getpass

OPENROUTER_API_KEY = getpass("Enter your OpenRouter API key (hidden): ")
OPENROUTER_BASE_URL = "https://openrouter.ai/api/v1"

from langchain_openai import ChatOpenAI
from langchain.prompts import PromptTemplate
from langchain.chains import LLMChain

# OpenRouter model configuration
model = ChatOpenAI(model="openai/gpt-4o-mini", temperature=0, seed=42, api_key=OPENROUTER_API_KEY, base_url=OPENROUTER_BASE_URL)
```

### Step 2: Create a `PromptTemplate`
A template separates intent from content. Keep placeholders explicit and typed in your head (e.g., `{topic}` is a short noun phrase).


```python
template = (
    "You are a concise tech writer. "
    "Explain {topic} to a junior developer in 4 sentences or fewer."
)
prompt = PromptTemplate.from_template(template)
print(prompt.format(topic="vector databases"))
```

### Step 3: Wrap into an `LLMChain`
`LLMChain` is glue: it takes your prompt, injects variables, calls the model, returns text. Keep it single-purpose.


```python
chain = LLMChain(llm=model, prompt=prompt)

result = chain.run({"topic": "LangChain"})
print(result)
```

### Step 4: Make it parameterized and reusable
We keep the interface minimal: a function that takes `topic` and returns text.

Note: prefer small, pure functions for testability and reuse.


```python
def explain_topic(topic: str) -> str:
    """Return a concise explanation for the given topic using the chain."""
    return chain.run({"topic": topic})

print(explain_topic("vector embeddings"))
```

### Step 5: A few pointers for your future projects...
- **Determinism**: `temperature=0`, `seed=42` to improve reproducibility.
- **Safety**: Keep prompts explicit and concise; avoid sensitive data.
- **Observability**: Print or log inputs/outputs while developing.
- **Separation**: Put keys in `.env` and load with `dotenv`.

You now have a minimal LangChain project: template → chain → function. Perfect base for adding Tools/Memory next.
