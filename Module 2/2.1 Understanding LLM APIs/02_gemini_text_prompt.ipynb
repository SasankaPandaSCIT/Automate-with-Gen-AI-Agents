### Step 1: Set your Openrouter API key (environment variable)
Never hardcode keys in notebooks. Use environment variables or a `.env` file.

- Create a file named `.env` next to this notebook with a line like:
  - `OPENROUTER_API_KEY=sk-...`
- Or export it in your terminal before launching Jupyter:
  - `export OPENROUTER_API_KEY=sk-...`

Run the next cell to load the key from environment.


```python
import os
from dotenv import load_dotenv

# Load environment variables from a .env file if present
load_dotenv()

# Load OpenRouter API key
openrouter_key = os.getenv("OPENROUTER_API_KEY")

# Quick sanity check (prints only whether it exists)
has_key = bool(openrouter_key)

print("OpenRouter key loaded:", has_key)
```

### Step 2: Make your first Gemini text request
- We'll use a small, fast model.
- Send one short prompt.
- Print the text output.


```python
from openai import OpenAI

client = OpenAI(
    api_key=os.getenv("OPENROUTER_API_KEY"),
    base_url="https://openrouter.ai/api/v1"
)

result = client.chat.completions.create(
    model="google/gemini-2.5-flash",
    messages=[
        {
            "role": "user",
            "content": "Say hello in one short sentence."
        }
    ]
)

print(result.choices[0].message.content)
```

### Step 3: Adjust randomness (temperature)
Try two temperatures and compare tone/variety.


```python
result = client.chat.completions.create(
    model="google/gemini-2.5-flash",
    messages=[
        {
            "role": "user",
            "content": "Give one quick mindfulness tip."
        }
    ],
    temperature=0.2
)

print("Temperature 0.2:\n", result.choices[0].message.content)


result = client.chat.completions.create(
    model="google/gemini-2.5-flash",
    messages=[
        {
            "role": "user",
            "content": "Give one quick mindfulness tip."
        }
    ],
    temperature=0.9
)

print("\nTemperature 0.9:\n", result.choices[0].message.content)
```
