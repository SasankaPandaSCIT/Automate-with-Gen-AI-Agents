### Step 1: Set your Openrouter API key (environment variable)
Never hardcode keys in notebooks. Use environment variables or a `.env` file.

- Create a file named `.env` next to this notebook with a line like:
  - `OPENROUTER_API_KEY=sk-...`
- Or export it in your terminal before launching Jupyter:
  - `export OPENROUTER_API_KEY=sk-...`

Run the next cell to load the key from environment.

```python
import getpass

# Enter your OpenRouter API key when prompted
openrouter_key = getpass.getpass("Enter your OpenRouter API key: ")

# Create the .env file
with open(".env", "w") as f:
    f.write(f"OPENROUTER_API_KEY={openrouter_key}\n")

print("OpenRouter API key saved to .env")
```

```python
import os
from dotenv import load_dotenv

# Load environment variables from a .env file if present
load_dotenv()

# Load OpenRouter API key
openrouter_key = os.getenv("OPENROUTER_API_KEY")

# Quick sanity check (prints only whether it exists)
has_key = bool(openrouter_key)

print("OpenRouter key loaded:", has_key)
```

### Step 2: Your first chat completion
We will:
- Create a simple system instruction
- Send a user message
- Read the model's reply

Stick to a tiny prompt. Short inputs = fast demos.


```python
from openai import OpenAI
import os

client = OpenAI(
    api_key=os.getenv("OPENROUTER_API_KEY"),
    base_url="https://openrouter.ai/api/v1"
)

response = client.chat.completions.create(
    model="openai/gpt-5",
    messages=[
        {"role": "system", "content": "You are a helpful and friendly assistant."},
        {"role": "user", "content": "Say hi in one short sentence."},
    ],
    temperature=1,
)

print(response.choices[0].message.content)
```

### Step 3: Tweak creativity (temperature)
- `temperature` controls randomness.
- Lower = more predictable, higher = more creative.

Try 0.2 vs 0.9 and observe the tone change.


```python
response = client.chat.completions.create(
    model="gpt-4o",
    messages=[
        {"role": "system", "content": "You are a concise assistant."},
        {"role": "user", "content": "Recommend one quick productivity tip."},
    ],
    temperature=0.2,
)
print("Temperature 0.2:\n", response.choices[0].message.content)

response = client.chat.completions.create(
    model="gpt-4o",
    messages=[
        {"role": "system", "content": "You are a concise assistant."},
        {"role": "user", "content": "Recommend one quick productivity tip."},
    ],
    temperature=0.9,
)
print("\nTemperature 0.9:\n", response.choices[0].message.content)
```
